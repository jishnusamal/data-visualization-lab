```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
students_df = pd.read_csv("Students.csv")

top_student = students_df.loc[students_df["Marks"].idxmax()]
print("Student with highest marks:")
print(top_student)

passed_students = students_df[students_df["Result"] == "Pass"]
passed_students.to_csv("PassedStudents.csv", index=False)

print("\nPassed students written to PassedStudents.csv")
display(passed_students)
```
