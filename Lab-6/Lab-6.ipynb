```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

```python
mtdf = pd.read_csv("mtcars.csv")
mtdf
```

```python
q1, ax1 = plt.subplots(figsize=(6, 4))
sns.boxplot(data=mtdf, x="am", y="mpg", ax=ax1)
ax1.set_title("Transmission Type vs MPG")
plt.show()
q1.savefig("ttvsmpg.png", dpi=300, bbox_inches="tight")

q2, ax2 = plt.subplots(figsize=(6, 4))
ax2.scatter(x=mtdf["hp"], y=mtdf["mpg"])
ax2.set_title("HP vs MPG")
ax2.set_xlabel("HP")
ax2.set_ylabel("MPG")
plt.show()
q2.savefig("hpvsmpg.png", dpi=300, bbox_inches="tight")

q2, ax3 = plt.subplots(figsize=(6, 4))
ax3.scatter(x=mtdf["wt"], y=mtdf["mpg"])
ax3.set_title("Weight vs Mileage")
ax3.set_xlabel("Weight")
ax3.set_ylabel("MPG")
plt.show()
q2.savefig("wtvsmpg.png", dpi=300, bbox_inches="tight")
```
