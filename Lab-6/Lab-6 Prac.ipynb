```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

```python
mtdf = pd.read_csv("mtcars.csv")
mtdf
```

```python
plt.figure()

plt.plot(mtdf["model"], mtdf["cyl"], marker='*')
plt.xlabel("Model")
plt.ylabel("Cylinders")

plt.bar(mtdf["model"], mtdf["mpg"])
plt.xlabel("Model")
plt.ylabel("Mpg")
```

```python
plt.hist(mtdf["mpg"], bins=10)
plt.xlabel("Model")
plt.ylabel("Cylinders")
```

```python
sns.boxplot(x=mtdf["mpg"], color='cyan')
# plt.xlabel("Model")
plt.xlabel("Cylinders")
```

```python
plt.scatter(mtdf["model"], mtdf["mpg"])
plt.xlabel("Model")
plt.ylabel("Mpg")
```

```python
shareholders = ["X", "Investors"]
equity = [60, 40]
fig = plt.figure()
plt.pie(equity, labels=shareholders, autopct="%d%%")
plt.show()
fig.savefig("Shares.png")
```
