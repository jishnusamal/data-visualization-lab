```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

```python
df = pd.read_csv("ipl batting.csv")
df
```

```python
print(f"Player with highest runs: {df.loc[df["Runs"].idxmax(), "Player"]} \n")

print(f"Player with highest strike rate: {df.loc[df["StrikeRate"].idxmax(), "Player"]} \n")

print(f"Player which has the highest average: {df.loc[df["Average"].idxmax(), "Player"]}\n")

print(f"Player which has the highest centuries: {df.loc[df["Centuries"].idxmax(), "Player"]}\n")

print(f"Total Runs: {df["Runs"].sum()} \n")

print(f"Average Strike Rate: {round(df["StrikeRate"].mean(), 2)} \n")

print(f"Total Centuries: {df["Centuries"].sum()} \n")

df["Runs per Match"] = round(df["Runs"]/df["Matches"], 2)
display(df)

plt.bar(df["Player"], df["Runs"])
plt.xticks(rotation=90)
plt.xlabel("Player")
plt.ylabel("Runs")
plt.title("Runs vs Players")
plt.savefig("img1.png")
plt.show()

plt.barh(df["Player"], df["StrikeRate"])
plt.xlabel("StrikeRate")
plt.ylabel("Player")
plt.title("Strike Rate vs Players")
plt.savefig("img2.png")
plt.show()

plt.pie(df["Runs"], labels=df["Player"], autopct="%.1f%%")
plt.title("Runs Distribution")
plt.savefig("img3.png")
plt.show()
```
